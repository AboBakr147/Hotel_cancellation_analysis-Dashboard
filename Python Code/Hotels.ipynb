## check outliers

```python
df.describe()
```

```python
# الأعمدة اللي فيها أوتلايرز قوية
cols_with_outliers = [
    'adr',
    'lead_time',
    'stays_in_week_nights',
    'stays_in_weekend_nights',
    'adults',
    'children',
    'babies',
    'days_in_waiting_list',
    'previous_bookings_not_canceled',
    'previous_cancellations'
]

# رسم شبكة من البوكس بلوتس
df[cols_with_outliers].plot(kind='box',subplots=True , layout=(5,2),figsize=(15,12))
plt.tight_layout()
plt.show()
```

```python
def count_outliers(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return ((series < lower) | (series > upper)).sum()

# جدول فيه عدد الأوتلايرز لكل عمود عددي
outlier_counts = {}
for col in df.select_dtypes(include=np.number).columns:
    outlier_counts[col] = count_outliers(df[col])

outlier_summary = pd.DataFrame.from_dict(outlier_counts, orient='index', columns=['Outlier_Count'])
print(outlier_summary.sort_values(by='Outlier_Count', ascending=False))
```

### If I delete all the outliers, I will lose a lot of data, so I will use capping

```python
df['adults'] = df['adults'].clip(upper=10)       # أي قيمة أكبر من 10 تتحول لـ 10
df['children'] = df['children'].clip(upper=5)    # أي قيمة أكبر من 5 تتحول لـ 5
df['babies'] = df['babies'].clip(upper=3)        # أي قيمة أكبر من 3 تتحول لـ 3
```
